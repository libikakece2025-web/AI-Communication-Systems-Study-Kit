```python
L0pppp
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 1000)

# Message signal
fm = 5
message = np.cos(2 * np.pi * fm * t)

# Carrier signal
fc = 50
carrier = np.cos(2 * np.pi * fc * t)

# Amplitude Modulation
modulation_index = 0.7
am_signal = (1 + modulation_index * message) * carrier

plt.figure(figsize=(10, 5))
plt.plot(t, am_signal, label="AM Signal")
plt.plot(t, message, label="Message Signal")
plt.title("AI Communication Study Kit - AM Simulation")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Time values
t = np.linspace(0, 1, 5000)

# Signal parameters
fm = 5
fc = 50
beta = 5

# Frequency Modulation signal
fm_signal = np.cos(
    2 * np.pi * fc * t
    + beta * np.sin(2 * np.pi * fm * t)
)

# Plot the FM signal
plt.figure(figsize=(10, 4))
plt.plot(t, fm_signal, label="FM Signal")
plt.title("Frequency Modulation (FM)")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Time settings
t = np.linspace(0, 1, 5000)

# Carrier frequency
fc = 20

# Digital data: 1 = signal ON, 0 = signal OFF
bit_rate = 5
bits = np.array([1, 0, 1, 1, 0])

samples_per_bit = len(t) // len(bits)
data = np.repeat(bits, samples_per_bit)
data = np.pad(data, (0, len(t) - len(data)), mode="edge")

# Carrier signal
carrier = np.cos(2 * np.pi * fc * t)

# ASK signal
ask_signal = data * carrier

# Plot
plt.figure(figsize=(12, 5))

plt.subplot(2, 1, 1)
plt.plot(t, data)
plt.title("Digital Input Bits")
plt.ylabel("Bit Value")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(t, ask_signal)
plt.title("ASK Modulated Signal")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()
```
